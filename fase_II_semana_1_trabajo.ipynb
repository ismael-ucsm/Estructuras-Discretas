```python
from pyDatalog import pyDatalog
```

```python
# Ejercicio 1
# Declaramos las variables X e Y (deben empezar con mayúscula)
pyDatalog.create_terms('X,Y')
```

```python
# Ejercicio 2
# a) Pregunta si X es igual a 1
print(X == 1)
```

```python
# b) Consulta con dos criterios combinados con '&'
print((X == True) & (Y == False))
```

```python
# Ejercicio 3
# Consulta con una contradicción: X no puede ser True y False al mismo tiempo
# El resultado es un conjunto vacío, ya que no existe ningún valor de X que satisfaga ambas condiciones
print((X == True) & (X == False))
```

```python
# Ejercicio 4
# Las variables también pueden representar cadenas de texto.
# Pide un nombre por teclado y arma un saludo con la primera letra de ese nombre
print((X == input('Please enter your name : ')) & (Y == 'Hello ' + X[0]))
```

```python
# Ejercicio 5
# Consulta con Y == 1 y a la vez Y == X + 1.
# Como X nunca fue definida pyDatalog no puede calcular X + 1, por lo que la solución es vacía.
print((Y == 1) & (Y == X + 1))
```

```python
# Ejercicio 6
# Aquí X sí está vinculada (X == 1), por lo que Y == X + 1 se puede evaluar.
# Como también se pide Y == 1, y X + 1 = 2, las condiciones son incompatibles entonces el resultado es un conjunto vacío.
# Si probamos con X == 0 sí habría solución.
print((X == 1) & (Y == 1) & (Y == X + 1))
```

```python
# Ejercicio 7
# X toma el valor (1,2,3) y Y toma el tercer elemento de X (índice 2)
print((X == (1, 2) + (3,)) & (Y == X[2]))
```

```python
# Ejercicio 8
# Se pueden usar funciones de Python dentro de las expresiones siempre que se creen como términos lógicos con create_terms.
def twice(a):
    return a + a

pyDatalog.create_terms('twice')
print((X == 1) & (Y == twice(X)))
```

```python
# Ejercicio 9
# Se pueden usar funciones de librerías estándar, como math.sqrt, dentro de las consultas.
pyDatalog.create_terms('math,Z')
print((X == 2) & (Y == math.sqrt(X)))
# Ejemplo de código
# (suma recursiva de los dígitos de un número)
def suma_digito(n):
    if n < 10:
        return n
    else:
        return n % 10 + suma_digito(n // 10)

print(suma_digito(123))
```

```python
# Ejercicio 11
# a) El resultado es un conjunto de posibles soluciones, en orden aleatorio.
# El atributo .data permite acceder a ese resultado como una lista de tuplas.
print(X.in_(range(5)).data)
print(X.in_(range(5)) == set([(0,), (1,), (2,), (3,), (4,)]))

# b) Después de una consulta, la variable guarda todos sus valores posibles .data devuelve la lista completa, .v() devuelve el primer valor,
# y '>=' extrae el primer valor de la consulta hacia la variable.
print("Data : ", X.data)
print("First value : ", X.v())
print("Extraction of first value of X: ", X.in_(range(5)) >= X)

# c) El resultado de una consulta se puede filtrar usando '&'.
# Aquí se queda solo con los valores de X menores a 2.
print(X.in_(range(5)) & (X < 2))
```

```python
# Ejercicio 12
# Los bucles se pueden anidar fácilmente combinando varios in_().
# Esta consulta devuelve todas las combinaciones de X e Y entre 0 y 4,
# donde Z es la suma de ambas y además Z es menor a 3.
print(X.in_(range(5)) & Y.in_(range(5)) & (Z == X + Y) & (Z < 3))
```
